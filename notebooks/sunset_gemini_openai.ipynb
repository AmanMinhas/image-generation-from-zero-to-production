# Generate a sunset image with Gemini through Chat Completions

This notebook uses the OpenAI Python SDK with Google's [Gemini-compatible Chat Completions endpoint](https://ai.google.dev/gemini-api/docs/openai). It requests an image from `gemini-2.5-flash-image`, saves it in `notebooks/outputs/`, and displays it inline.

Google's Gemini Developer API documentation demonstrates image generation through the Images API, but does not document image output from Chat Completions. [Google Cloud's separate OpenAI-compatible endpoint](https://docs.cloud.google.com/gemini-enterprise-agent-platform/models/migrate/openai/examples#image_generation) demonstrates a Chat Completions image response in `message.audio.data`. This notebook accepts that response shape and image data URLs in message parts. The Gemini Developer API behavior needs a live run to confirm.

Install the project dependencies with `UV_CACHE_DIR=.uv-cache uv sync`, select the project's `.venv` kernel, and run the cells in order. An API call may incur charges.

```python
import base64
import os
from getpass import getpass
from pathlib import Path

from IPython.display import Image, display
from openai import OpenAI

api_key = os.environ.get("GEMINI_API_KEY") or getpass("Gemini API key: ")
meta_api_key = os.environ.get("MODEL_API_KEY") or getpass("Model API key: ")
if not api_key:
    raise ValueError("Set GEMINI_API_KEY or enter a Gemini API key when prompted.")
if not meta_api_key:
    raise ValueError("Set MODEL_API_KEY or enter a Model API key when prompted.")

client = OpenAI(
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)
```

```python
MODEL = "gemini-2.5-flash-image"
# MODEL = "gemini-3.1-flash-image"
PROMPT = (
    "A tranquil sunset over the ocean, with a glowing orange sun just above the horizon, "
    "soft pink and violet clouds, gentle reflections across the water, and a dark "
    "silhouette of a distant coastline. Photorealistic landscape, no text."
)

response = client.interaction.create(
    model=MODEL,
    input=PROMPT
)

# response = client.images.generate(
#     model=MODEL,
#     prompt=PROMPT,
#     response_format='b64_json',
#     n=1,
# )

# for image_data in response.data:
#   generated_image = base64.b64decode(image_data.b64_json)
#   display(Image(data=generated_image, width=300))
```
